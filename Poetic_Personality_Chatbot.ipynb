## 3. Poetic Response Function (Prompt Engineering)

This is the heart of the chatbot's personality. Instead of sending the user's raw message to Gemini, we wrap it inside a **system-style instruction prompt** that:

- Tells the model to always respond in poetic verse (never plain prose)
- Asks for imagery, rhythm, and emotional resonance
- Constrains the length so responses stay tight and evocative (4-8 lines)
- Asks the model to reflect the *feeling* behind the user's message, not just the literal words

This is prompt engineering in action: we are shaping the model's behavior purely through instructions, without any fine-tuning.

````python
import os
from openai import OpenAI

client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=os.getenv("OPENROUTER_API_KEY")
)


def get_poetic_response(user_message: str) -> str:
    """
    Sends the user's message to OpenRouter wrapped in a poetic-persona prompt,
    and returns the model's poetic reply as a string.
    """

    poetic_prompt = f"""
You are a thoughtful, lyrical poet-companion. A person has just shared
the following thought or feeling with you:

\"{user_message}\"

Respond ONLY in poetic verse (never plain prose, never bullet points).
Your response should:
- Be 4 to 8 lines long
- Use vivid imagery and gentle rhythm (rhyme is welcome but not required)
- Emotionally reflect the meaning or feeling behind what they said
- Feel personal, warm, and profound - like a poem written just for them
- Avoid explanations, titles, or any text that is not part of the poem itself
"""

    response = client.chat.completions.create(
        model="google/gemma-4-26b-a4b-it:free",
        messages=[
            {"role": "system", "content": poetic_prompt},
            {"role": "user", "content": user_message}
        ]
    )
    return response.choices[0].message.content.strip()
````

## 4. Typing Animation Effect

To make the chatbot feel alive - as if a poet is composing verses in real time - we print the poem character by character with a small delay between each one, rather than dumping the whole response on screen at once.

````python
import sys
import time


def type_out(text: str, delay: float = 0.02):
    """
    Prints text one character at a time to simulate a live typing effect.
    'delay' controls the speed (seconds between characters).
    """
    for char in text:
        sys.stdout.write(char)
        sys.stdout.flush()
        time.sleep(delay)
    print()  # newline once the poem has finished 'typing'
````

## 5. Chatbot Loop

Finally, we tie everything together into an interactive loop:

1. The user types a message
2. We send it to `get_poetic_response()` to generate a poem
3. We display the poem using the `type_out()` animation
4. The loop repeats until the user types `exit` or `quit`

Run the cell below and start chatting with your Poetic Personality Chatbot!

````python
import os
import time
from openai import OpenAI

# Load API key from environment variable (recommended)
# Set it first: export OPENROUTER_API_KEY="your-key-here"
client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=os.getenv("OPENROUTER_API_KEY")
)

system_prompt = (
    "You are a poetic chatbot. Always respond in poetic verse, never in plain prose. "
    "Incorporate rich imagery, rhythm, and emotional resonance. "
    "Keep responses tight and evocative, strictly between 4 to 8 lines. "
    "Reflect the deep feeling behind the user's message, rather than just repeating literal words."
)


def type_out(text, delay=0.03):
    for char in text:
        print(char, end='', flush=True)
        time.sleep(delay)
    print()


def get_poetic_response(user_message: str) -> str:
    completion = client.chat.completions.create(
        model="google/gemma-4-26b-a4b-it:free",
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_message}
        ]
    )
    return completion.choices[0].message.content


def run_chatbot():
    print('🌙 Poetic Personality Chatbot 🌙')
    print('Share a thought or feeling, and I will turn it into verse.')
    print("Type 'exit' or 'quit' to end our conversation.\n")

    while True:
        user_message = input('You: ')

        if user_message.strip().lower() in ('exit', 'quit'):
            print('\nPoet: Until we speak again, may your days read like poetry. 🌸')
            break

        if not user_message.strip():
            continue

        try:
            poem = get_poetic_response(user_message)
            print('\nPoet:')
            type_out(poem)
            print()
        except Exception as e:
            print(f'⚠️ Something went wrong while generating your poem: {type(e).__name__}: {e}')
            print()


if __name__ == "__main__":
    run_chatbot()
````
